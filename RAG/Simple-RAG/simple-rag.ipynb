## simple_rag.py
 
A BEGINNER-FRIENDLY starting point for a RAG pipeline.
 
Just two things happen here:
  1. Read a text file
  2. Split it into small overlapping chunks
 
No classes, no fancy abstractions -- just plain functions so it's easy
to read top to bottom. We'll build on this once it makes sense.


```python
def read_text_file(file_path):
    with open(file_path , 'r', encoding='utf-8') as f:
        text = f.read()
    return text
```

```python
text = read_text_file("sample.txt")
print(text)
```

```python
def chunk_text(text , chunk_size = 200, overlap = 50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        start = end - overlap
    return chunks
```

```python
text = read_text_file("sample.txt")
chunks = chunk_text(text)
print(chunks)
```

```python
if __name__ == "__main__":
    text = read_text_file("sample.txt")
    print(f"Read {len(text)} Charecters from sample.txt\n")
    chunks = chunk_text(text)
    print(f"Splitting into {len(chunks)} chunks:\n")
    for i,chunk in enumerate(chunks):
        print(f"--- Chunk {i} ({len(chunk)} chars)---")
        print(chunk)
        print()
```
