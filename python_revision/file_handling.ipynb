# File Handling
The key function for working with files in Python is the open() function.

The `open()` function takes two parameters; filename, and mode.

There are four different methods (modes) for opening a file:

- "r" - Read - Default value. Opens a file for reading, error if the file does not exist

- "a" - Append - Opens a file for appending, creates the file if it does not exist

- "w" - Write - Opens a file for writing, creates the file if it does not exist

- "x" - Create - Creates the specified file, returns an error if the file exists

```python
# open file in read mode
f = open("sample.txt",'r')
print(f.read())
f.close()
```

# Using the with statement
You can also use the with statement when opening a file:
- Then you do not have to worry about closing your files, the with statement takes care of that.

```python
with open('sample.txt','r') as f:
    print(f.read())
```

```python
# Read first 5 characters of the file
with open('sample.txt','r') as f:
    print(f.read(5))
```

```python
# Read one line
with open('sample.txt','r') as f:
    print(f.readline())
```

```python
# loop on the file line by line
with open('sample.txt','r') as f:
    for x in f:
        print(x)
```

# Write to an Existing File
To write to an existing file, you must add a parameter to the open() function:

- "a" - Append - will append to the end of the file
- "w" - Write - will overwrite any existing content

```python
# open the file and append one line to it
with open('sample.txt' , 'a') as f:
    f.write("\nNew Lines have been added")

# open the file after appending new line
with open('sample.txt', 'r') as f:
    print(f.read())
```

```python
# open the file and overwrite
with open('sample.txt','w') as f:
    f.write("The file has been appended")

# open the file and read
with open('sample.txt','r') as f:
    print(f.read())
```

# Create a New File
To create a new file in Python, use the open() method, with one of the following parameters:

- "x" - Create - will create a file, returns an error if the file exists
- "a" - Append - will create a file if the specified file does not exists
- "w" - Write - will create a file if the specified file does not exists

```python
# Create new file
f = open('new_file.txt','x')
f.close()
```

# Delete a File
To delete a file, you must import the `OS` module, and run its `os.remove()` function:

```python
import os
os.remove('new_file.txt')
```

```python
# check if file exist then delete it
import os
if os.path.exists('new_file.txt'):
    os.remove('new_file.txt')
else:
    print("File is not exist")
```
