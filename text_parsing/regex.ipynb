## How to Use the Python re Module with RegEx
- Python provides some methods you can use to work with regular expressions such as `match()`, `search()`, `findall()`, `split()`, and `sub()`.

- `re.match(pattern, string)` # to use the match method
- `re.findall(pattern, string)` # to use the findall method 
- `re.sub(pattern, string)` # to use the sub method
- `re.search(pattern, string)` # to use the search method 
- `re.split(pattern, string)` # to use the split method

## Common Regex Patterns

| Regex   | Meaning                      | Example Match                |
|----------|-----------------------------|------------------------------|
| `.`      | Any character except newline | `a`, `1`, `@` |
| `^`      | Start of string | `^Hello` |
| `$`      | End of string | `world$` |
| `*`      | 0 or more occurrences | `ab*c` → `ac`, `abc`, `abbc` |
| `+`      | 1 or more occurrences | `ab+c` → `abc`, `abbc` |
| `?`      | 0 or 1 occurrence | `colou?r` |
| `{n}`    | Exactly n times | `\d{4}` |
| `{n,}`   | At least n times | `\d{2,}` |
| `{n,m}`  | Between n and m times | `\d{2,4}` |
| `[]`     | Character class | `[abc]` |
| `[^]`    | Negated character class | `[^abc]` |
| `\|`     | OR operator | `cat\|dog` |
| `()`     | Grouping | `(abc)+` |

## Character Classes

| Regex | Meaning |
|--------|---------|
| `\d` | Digit (`0-9`) |
| `\D` | Non-digit |
| `\w` | Word character (`a-z`, `A-Z`, `0-9`, `_`) |
| `\W` | Non-word character |
| `\s` | Whitespace |
| `\S` | Non-whitespace |
| `[a-z]` | Lowercase letters |
| `[A-Z]` | Uppercase letters |
| `[0-9]` | Digits |
| `[a-zA-Z]` | Any letter |
| `[a-zA-Z0-9]` | Alphanumeric |
| `(?i)`  |  Ignore case of letter|

## Anchors

| Regex | Meaning |
|--------|---------|
| `^` | Start of string |
| `$` | End of string |
| `\b` | Word boundary |
| `\B` | Not a word boundary |

```python
import re
my_text = "freeCodeCamp"
my_regex = "^freecodecamp$"

res = re.match(my_regex,my_text)
print(res)
# Returns none as regex all small leters while original text contains capital letters
```

```python
# to ignore letter case we have to methods:
# 1- using (?i)
# 2- using re.IGNORECASE
my_text = "freeCodeCamp"
my_regex = "(?i)^freecodecamp$"

res1 = re.match(my_regex,my_text)
print(res1)
```

```python
my_text = "freeCodeCamp"
my_regex = "^freecodecamp$"

res2 = re.match(my_regex,my_text,re.IGNORECASE)
print(res2)
```

```python
# RegEx Example Using the search() Method

my_text = 'Every Friday, we have a standup meeting. The only reason why we might not have a meeting on a Friday is public holiday. That Friday, we talk about what we did in the previous week, and what we are going to do in the week starting from that Friday.'
my_regex = 'friday'
res3 = re.search(my_regex,my_text,re.IGNORECASE)
print(res3)

# to get the start index of occurence we can use .start() and end using .end()
print("The first occurence is located at index", res3.start())
print("The occurence end at index", res3.end())
```

```python
# RegEx Example Using the findall() Method
my_txt = 'Every Friday, we have a standup meeting. The only reason why we might not have a meeting on a Friday is public holiday. That Friday, we talk about what we did in the previous week, and what we are going to do in the week starting from that Friday.'

my_regex = 'friday'
res = re.findall(my_regex, my_txt, re.IGNORECASE)

print(res) # Output: ['Friday', 'Friday', 'Friday', 'Friday']
```
